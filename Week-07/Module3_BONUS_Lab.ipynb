#### Creating Graphs: Group by Product and Region (Sum of Units & Revenue)

```python
# Data
data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
    'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]
}

df_sales = pd.DataFrame(data)
df_sales
```

```python
#### Applying revenue data from 2C to be defined 
```

```python
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
# df_sales['Column_Name'] Accesses or creates a column in the DataFrame
# (=) Stores the result as a new or updated column

#df_sales['Revenue'] accesses/creates a column in the DataFrame
# df_sales['Units_Sold'] * df_sales['Price'] multiplies values row-by-row 
```

```python
# df_sales['Units_Sold'] selects the Units_Sold column from the DataFrame
# .sum() adds up all values in that column to get the total
total_units_sold = df_sales['Units_Sold'].sum()
total_revenue = df_sales['Revenue'].sum()
# df_sales['Revenue'] selects the Revenue column
# .sum() calculates the total revenue across all rows
print("Total Units Sold:", total_units_sold)
print("Total Revenue:", total_revenue)
```

#### Grouped Bar Chart (Units Sold)

```python
# Group the data
grouped_product_region = df_sales.groupby(['Product', 'Region'])[['Units_Sold', 'Revenue']].sum()

# Reshape data for plotting (Units Sold)
units_sold_unstacked = grouped_product_region['Units_Sold'].unstack()

# Create bar chart
units_sold_unstacked.plot(kind='bar', figsize=(10, 6))

# Customize
plt.title('Units Sold by Product and Region')
plt.xlabel('Product')
plt.ylabel('Units Sold')
plt.xticks(rotation=45)
plt.legend(title='Region')

plt.tight_layout()
plt.show()
```

### Grouped Bar Chart (Revenue)

```python
# Reshape data for plotting (Revenue)
revenue_unstacked = grouped_product_region['Revenue'].unstack()

# Create bar chart
revenue_unstacked.plot(kind='bar', figsize=(10, 6))

# Customize
plt.title('Revenue by Product and Region')
plt.xlabel('Product')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.legend(title='Region')

plt.tight_layout()
plt.show()
```

```python
# Analysis:
# Units Sold by Product and Region:
#The chart shows how sales volume varies across 
# products and regions.Some products clearly outperform
# others in specific regions, indicating regional demand differences.
# If one region consistently has higher bars across products, that region
# is likely your strongest market in terms of volume.
```

```python
# Analysis:
# Revenue by Product and Region:
# This chart shows financial performance, not just volume.
# A product with lower units sold but high revenue may have a
# higher price point and/or Better margins
```

```python
# Summary:
# The Units Sold graph highlights demand and popularity
# The Revenue graph highlights profitability and value
# Together, they help identify Top-performing products,
# High-value markets,and Opportunities for pricing and expansion.
```
