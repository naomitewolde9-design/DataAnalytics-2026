## Permutations

### *refers to an arrangement of elements in a specific order, where the order matters*

```python
# current output for r_4
print("Original r_4:\n", r_4)
```

#### Shuffle r_4

```python
np.random.shuffle(r_4)

# Print shuffled r_4
print("Shuffled r_4:\n", r_4)
```

```python
# After using np.random.shuffle, the rows of r_4 were rearranged,
# but the values within each row stayed in the same order.
# This is because shuffle only rearranges elements along the first axis
# (rows for a 2D array), not the individual elements inside each row.
```

#### Create a new shuffled version without changing the original

```python
new_shuffle = np.random.permutation(r_4)

print("New shuffled array:\n", new_shuffle)
print("\nCurrent r_4 (unchanged after permutation):\n", r_4)
```

```python
# np.random.shuffle changes the original array by rearranging its elements,
# while np.random.permutation creates a new shuffled array without changing
# the original. You would use shuffle when you want to permanently change the order of an array,
# and permutation when you want a shuffled copy but still keep the original data unchanged.
```

# Teamwork

```python
# Ali: Monday → Friday
ali = np.array([5, 4, 6, 4, 2])
print("Ali:", ali)
```

```python
# Ben missed Monday → use 0
ben = np.array([0, 5, 5, 3, 4])
print("Ben:", ben)
```

```python
# Carla missed Wednesday → use 0
carla = np.array([3, 6, 0, 3, 4])
print("Carla:", carla)
```

```python
# Combine into 2D array (rows = people, columns = days)
all_knocks = np.array([ali, ben, carla])
print("All Knocks:\n", all_knocks)
```

```python
# Total per person
total_per_person = np.sum(all_knocks, axis=1)

# Total for entire group
total_group = np.sum(all_knocks)

print("Total doors knocked per person:")
print(f"Ali: {total_per_person[0]}")
print(f"Ben: {total_per_person[1]}")
print(f"Carla: {total_per_person[2]}")

print(f"\nTotal doors knocked by the group: {total_group}")
```

```python
#Total knocks per day
daily_totals = np.sum(all_knocks, axis=0)

# Rolling total of knocks
rolling_knocks = np.cumsum(daily_totals)

# Estimate votes (1 vote per 2 knocks)
estimated_votes = rolling_knocks * 0.5

print("Estimated votes (rolling total by day):\n", estimated_votes)
```
