## Filtering a DataFrame

### FIlter age > 18 

```python
df_age_filtered = df[df["Age"] > 18] #df["Age"] > 18, keeps rows where age is above 18

df_age_filtered
```

### FIlter score between 50 ansd 100

```python
df_score_filtered = df[(df["Score"] > 50) & (df["Score"] < 100)] # & means AND

df_score_filtered
```

## Modifying a DataFrame

### Remove invalid Score rows

```python
df_cleaned = df[df["Score"] >= 0] #df[df["Score"] >= 0] removes invalid scores like -1

df_cleaned
```

### Remove the City column from the DataFrame

```python
df_cleaned = df_cleaned.drop(columns=["City"]) # drop(columns=[...]) removes columns
 
df_cleaned
```

### Replace all missing Age values with the mean Age of the DataFrame using fillna

```python
df_cleaned["Age"] = df_cleaned["Age"].fillna(df_cleaned["Age"].mean()) # fillna(mean) replaces missing values

df_cleaned
```

### Replace any Score value of 0 with the mean Score of the DataFrame using replace

```python
df_cleaned["Score"] = df_cleaned["Score"].replace(0, df_cleaned["Score"].mean()) # replace() swaps specific values (0 = mean score)

df_cleaned
```

### Remove any remaining rows that contain NaN values using the dropna function

```python
df_cleaned = df_cleaned.dropna() # dropna() removes any leftover missing data 

df_cleaned
```
