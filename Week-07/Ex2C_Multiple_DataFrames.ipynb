## Basic Aggregations

```python
data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
    'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]
}

df_sales = pd.DataFrame(data)
df_sales
```

```python
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
# df_sales['Column_Name'] Accesses or creates a column in the DataFrame
# (=) Stores the result as a new or updated column

#df_sales['Revenue'] accesses/creates a column in the DataFrame
# df_sales['Units_Sold'] * df_sales['Price'] multiplies values row-by-row 
```

```python
# df_sales['Units_Sold'] selects the Units_Sold column from the DataFrame
# .sum() adds up all values in that column to get the total
total_units_sold = df_sales['Units_Sold'].sum()
total_revenue = df_sales['Revenue'].sum()
# df_sales['Revenue'] selects the Revenue column
# .sum() calculates the total revenue across all rows
print("Total Units Sold:", total_units_sold)
print("Total Revenue:", total_revenue)
```

```python
# df_sales['Units_Sold'] selects the Units_Sold column from the DataFrame
# .mean() calculates the average (mean) of all values in that column
average_units_sold = df_sales['Units_Sold'].mean()
average_revenue = df_sales['Revenue'].mean()
# df_sales['Revenue'] selects the Revenue column
# .mean() computes the average revenue across all rows
print("Average Units Sold:", average_units_sold)
print("Average Revenue:", average_revenue)
```

## Grouping Data

### Group by Product (Total Units & Revenue)

```python
grouped_product = df_sales.groupby('Product')[['Units_Sold', 'Revenue']].sum() #groupby() splits data into groups based on column(s)/ sum() adds values within each group
grouped_product 

#groupby('Product') splits the DataFrame into groups based on unique values in the 'Product' column
# [['Units_Sold', 'Revenue']] selects only the columns we want to aggregate
# .sum() adds up all values within each group for those columns
```

### Group by Region (Average Units & Revenue)

```python
grouped_region = df_sales.groupby('Region')[['Units_Sold', 'Revenue']].mean() # mean() calculates averages within each group
grouped_region
# groupby('Region') groups the data based on each unique region
# [['Units_Sold', 'Revenue']] selects columns to calculate statistics on
# .mean() calculates the average (mean) value for each group
```

### Group by Product and Region (Sum of Units & Revenue)

```python
grouped_product_region = df_sales.groupby(['Product', 'Region'])[['Units_Sold', 'Revenue']].sum() # [['col1', 'col2']] → selects specific columns to work with
grouped_product_region
# groupby(['Product', 'Region']) creates multi-level grouping (hierarchical index)
# This means data is grouped by each unique combination of Product AND Region
# [['Units_Sold', 'Revenue']] selects the relevant numeric columns
# .sum() computes the total for each Product-Region combination
```

## Sorting and Ranking Data

### Sort by Units_Sold (Ascending)

```python
# sort_values() is used to sort a DataFrame by a specific column
# 'Units_Sold' is the column we are sorting by
# ascending=True sorts from smallest to largest
sorted_units = df_sales.sort_values(by='Units_Sold', ascending=True) # by=  Specifies which column to sort

sorted_units
```

### Sort by Revenue (Descending)

```python
# sort_values() sorts the data based on 'Revenue'
# ascending=False sorts from highest to lowest
sorted_revenue = df_sales.sort_values(by='Revenue', ascending=False) # ascending=False Sorts from highest to lowest

sorted_revenue
```

```python
# rank() assigns a rank to each value in the column
# ascending=False means highest value gets rank 1
# method='dense' makes sure that ranks have no gaps in-between 
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending=False, method='dense') 
#rank() Assigns a numerical ranking to values/method='dense'Prevents skipped ranking numbers (For exp: 1,2,3 instead of 1,2,4)

df_sales
```

## Merging DataFrames

### Create df_products

```python
# Define product data using a dictionary
# Keys become column names, values are the data for each column
product_data = {
    'Product': ['A', 'B', 'C'],
    'Category': ['Electronics', 'Furniture', 'Appliances']
}

# pd.DataFrame() Converts structured data into a table format
df_products = pd.DataFrame(product_data) 

df_products
```

### Merge DataFrames

```python
# pd.merge() combines two DataFrames based on a common column
# 'on="Product"' specifies the column both DataFrames share
df_merged = pd.merge(df_sales, df_products, on='Product')
# Default join = inner join (only matching values kept)
df_merged
```

```python
# groupby('Category') groups the merged data by product category
# ['Revenue'] selects the column we want to aggregate
# .sum() calculates total revenue for each category
revenue_by_category = df_merged.groupby('Category')['Revenue'].sum()

revenue_by_category
```
