```python
import numpy as np 
```

```python
import matplotlib.pyplot as plt
```

## Advanced Plotting Techniques

### Basic Bar Chart

```python
# Data
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

# Bar chart
plt.bar(products, sales)

plt.show()
```

### Customized Bar Chart

```python
# Data
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

# Customized bar chart
plt.bar(products, sales, color='skyblue', width=0.5)

plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')

plt.show()
```

### Basic Scatter Plot

```python
x = np.random.rand(50)
y = np.random.rand(50)

# Scatter plot
plt.scatter(x, y)

plt.show()
```

### Customized Scatter Plot

```python
x = np.random.rand(50)
y = np.random.rand(50)
sizes = np.random.rand(50) * 100  # marker sizes

# Scatter plot with size and colormap
plt.scatter(x, y, s=sizes, c=sizes, cmap='viridis')

plt.colorbar()  # show color scale

plt.title('Scatter Plot with Variable Sizes and Colors')
plt.xlabel('X values')
plt.ylabel('Y values')

plt.show()
```

## Subplots

### Create Subplots with Line Chart and Histogram (side by side)

```python
# Create figure with two subplots
plt.figure(figsize=(10, 5))

# First subplot: Sine wave
plt.subplot(1, 2, 1)  # 1 row, 2 columns, position 1
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)

plt.title('Sine Wave')
plt.xlabel('X values')
plt.ylabel('sin(X)')

# Second subplot: Histogram
plt.subplot(1, 2, 2)  # position 2
data = np.random.randn(1000)
plt.hist(data, bins=30)

plt.title('Histogram of Random Data')
plt.xlabel('Value')
plt.ylabel('Frequency')

# Adjusts the layout
plt.tight_layout()

# Display
plt.show()
```
