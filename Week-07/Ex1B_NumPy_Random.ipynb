# Exercise 1.B NumPy Random

```python
import numpy as np 
```

## Generating Random Numbers

### r1

```python
# Random ints between 1 and 20 (size = 2 int)
r_1 = np.random.randint(1, 21, size=2)
print(r_1)
```

### r2

```python
# Random int between 1 and 20 (1 row, 2 columns)
r_2 = np.random.randint(1, 21, size=(1, 2))
print(r_2)
```

### r3

```python
# Random floats between 0 and 1 (2 rows, 1 column)
r_3 = np.random.rand(2, 1)
print(r_3)
```

### r4

```python
# Random int between 20 and 100 (4 rows, 5 columns)
r_4 = np.random.randint(20, 101, size=(4, 5))
print(r_4)
```

### r5

```python
# Random floats between 0 and 1 (4 rows, 5 columns)
r_5 = np.random.rand(4, 5)
print(r_5)
```

## Generating Random Choices from an Array

#### Creating Students: 1-D array of 12 classmates in YUU

```python
students = np.array([
    "Lesley", "Corey", "Lorah", "Alondra",
    "Victoria", "Alexus", "Blake", "Tihitina",
    "Hana", "Charles", "Luis", "Smaya"
])

print(students)
```

```python
# Randomly select one student
call_on = np.random.choice(students)
print(call_on)
```

### Creating Breakout groups 

```python
# Random sampling into a 3x4 group (3, 4)
breakout_groups = np.random.choice(students, size=(3, 4))
print(f"Breakout Groups: \n" ,breakout_groups)
```

```python
# Comment:
# This is NOT an effective way to assign breakout groups because
# students can be selected more than once (duplicates),
# and some students might not be selected at all.
# This makes the grouping unfair and uneven.
```

### Create statement for 5 random selections of students 

```python
p_values = [0.0, 0.30, 0.10, 0.08,
            0.07, 0.07, 0.08, 0.08,
            0.07, 0.05, 0.05, 0.05]

call_on_weighted = np.random.choice(students, size=5, p=p_values)
print(f"Random Students: \n" ,call_on_weighted)
```

## Generating a Normal Distribution

#### Create a normal (bell curve) distribution

```python
# mean = 10, standard deviation = 5, size = (4, 5)

bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(f"Bell Curve: \n", bell_curve)
```
