## Creating Basic Plots

```python
# Create variables
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

#### Basic Line Plot

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# Plot
plt.plot(x, y)

# Display
plt.show()
```

#### Line Plot with Markers

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# Plot with markers
plt.plot(x, y, marker='o')

# Display
plt.show()
```

#### Customized Plot

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# Plot with markers
plt.plot(x, y, marker='o')

# Add labels and title
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

# Display
plt.show()
```

 ## Styling Plots

#### Dashed Red Line

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# Plot with dashed red line
plt.plot(x, y, linestyle='--', color='red')

plt.show()
```

#### Square Markers (Customized)

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# Plot with square markers
plt.plot(x, y, linestyle='--', color='red',
         marker='s', markersize=10, markerfacecolor='green')

plt.show()
```

#### Experiment with Different Markers and Colors

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# Plot 1: Triangle markers, blue line
plt.plot(x, y, marker='^', color='blue', label='Triangle markers')

# Plot 2: Diamond markers, purple line
plt.plot(x, y, marker='D', color='purple', label='Diamond markers')

plt.legend()
plt.show()
```
