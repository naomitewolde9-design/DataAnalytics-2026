```python
import numpy as np
```

## Array Basics 

```python
# Fruit Variables
fruits = ["apple", "banana", "cherry", "date"]
```

```python
# Create Numpy Array
arr_1 = np.array(fruits)
print(arr_1)
```

```python
# Checking Data Type 
print(arr_1.dtype)
```

```python
# Creat 2-D Array 
arr_2 = np.array([[10, 11, 12], 
                  [13, 14, 15]])

print(arr_2)
```

```python
# Transpose 
print(arr_2.T)
```

```python
# Checking to see if arr_2 changed 
print(arr_2)
```

```python
#No, this does not change arr_2 permanently. Just the format temporarily
#(.T) only returns a transposed view, it doesn't modify the original array.
```

```python
# Creating another 2-D Array 
arr_3 = np.array([[1, 3, 5],
                  [2, 4, 6]])
print(arr_3)
```

```python
# Using Slicing with an Array 
arr_4 = np.array([arr_3[0, 2], arr_3[1, 1]])

print(arr_4)
```

```python
# Creating 3-D Array (creates array of all 1's)
arr_5 = np.ones((2, 2, 2))

print(arr_5)
```
