### Exercise 2.E Using functions for analyzing data
This notebook analyzes a small sales dataset using Python functions.
Each function takes the full dataset as its first argument.

```python
sales_data = [
 ('Laptop', 'East', 1200.00),
 ('Phone', 'West', 850.00),
 ('Tablet', 'East', 400.00),
 ('Monitor', 'South', 650.00),
 ('Keyboard', 'West', 75.00),
 ('Laptop', 'South', 1150.00),
 ('Phone', 'East', 900.00),
 ('Tablet', 'West', 425.00),
 ('Monitor', 'East', 680.00),
 ('Keyboard', 'South', 80.00),
 ('Laptop', 'West', 1300.00),
 ('Phone', 'West', 820.00),
]
```

## Lab 1. Total sales by region

```python
def region_total(data, region):
    total = 0
    for product, r, amount in data:
        if r == region:
            total += amount
    return total
```

```python
print(region_total(sales_data, 'East'))
print(region_total(sales_data, 'West'))
print(region_total(sales_data, 'South'))
```

## Lab 2. Count of sales by region

```python
def region_count(data, region):
    return sum(1 for product, r, amount in data if r == region)
```

```python
print(region_count(sales_data, 'East'))
print(region_count(sales_data, 'West'))
print(region_count(sales_data, 'South'))
```

## Lab 3. Average sale amount by region

```python
def region_average(data, region):
    total = region_total(data, region)
    count = region_count(data, region)
    
    if count == 0:
        return 0
    return total / count
```

```python
print(region_average(sales_data, 'East'))
print(region_average(sales_data, 'West'))
print(region_average(sales_data, 'South'))
```

```python
print(region_average(sales_data, 'North'))
```

## Lab 4. Filter by sale amount

```python
def sales_above(data, threshold):
    return [record for record in data if record[2] > threshold]
```

```python
results = sales_above(sales_data, 500)
```

```python
for record in results:
    print(record)
```

## Lab 5. Top and bottom sale

```python
def top_sale(data):
    return max(data, key=lambda record: record[2])
```

```python
def bottom_sale(data):
    return min(data, key=lambda record: record[2])
```

```python
print(top_sale(sales_data))
print(bottom_sale(sales_data))
```

### RESULT SUMMARY 

The top sale was a Laptop in the West region for 1300.00
The lowest sale was a keyboard in the West region for 75.00

## Lab 6. Apply a discount

```python
def apply_discount(data, discount_pct):
    return list(map(lambda s: (s[0], s[1], s[2] * (1 - discount_pct)), data))
```

```python
discounted_sales = apply_discount(sales_data, 0.10)
```

```python
for record in discounted_sales:
    print(record)
```

### Trying 15% discount

```python
discounted_sales_15 = apply_discount(sales_data, 0.15)
```

```python
print("Original (first 3):")
for record in sales_data[:3]:
    print(record)
```

```python
print("\nAfter 15% discount (first 3):")
for record in discounted_sales_15[:3]:
    print(record)
```

## Lab 7. Unique regions

```python
def unique_regions(data):
    return sorted(set(region for product, region, amount in data))
```

```python
print(unique_regions(sales_data))
```

### Analysis
The unique_regions function helps find all distinct regions in the data.  
This is useful for grouping, filtering, or running analysis (like totals or averages)
for each region without hardcoding names.


## Lab 8. Region summary

```python

def region_summary(data, region):
    total = region_total(data, region)
    count = region_count(data, region)
    average = region_average(data, region)
    
    region_data = [record for record in data if record[1] == region]
    
    top = top_sale(region_data) if region_data else None
    
    # Print formatted report
    print(f"=== {region} Region ===")
    print(f"Records: {count}")
    print(f"Total Sales: ${total:,.2f}")
    print(f"Average Sale: ${average:,.2f}")
    
    if top:
        print(f"Top Sale: {top[0]} — ${top[2]:,.2f}")
    else:
        print("Top Sale: None")
```

```python
region_summary(sales_data, 'East')
region_summary(sales_data, 'West')
region_summary(sales_data, 'South')
```

## Lab 9. Full summary 

```python
for r in unique_regions(sales_data):
    region_summary(sales_data, r)
    print()  # blank line for readability
```
