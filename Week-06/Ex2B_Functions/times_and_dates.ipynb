## working with datetime functions

```python
import datetime
```

### Today 

```python
today = datetime.datetime.now() # gets the current date and exact time 
```

```python
print(today)
```

```python
formatted_date = today.strftime("Today is %A, %B %d, %Y") # gets the day of week, month name,day of month, year
```

```python
print(formatted_date)
```

### My birthday 

```python
birthday = datetime.date(2003, 2, 13)
```

```python
formatted_birthday = birthday.strftime("My birthday is %m/%d/%Y")
```

```python
print(formatted_birthday)
```

### Next 90 Days

```python
# calculating the date 90 days from today
ninety_d = today + datetime.timedelta(days=90)
```

```python
print(ninety_d)
```

```python
formatted_90_days = ninety_d.strftime("90 days from today is %B %d, %Y") # full (m) name,(d) day, and full (y)

print(formatted_90_days)
```

### Time for dinner 

```python
#define dinner time (hour in 24hr form)
dinner_time = datetime.time(18, 30)

print(dinner_time)
```

```python
# format dinner time as HH:MM PM
formatted_dinner = dinner_time.strftime("Let’s meet for dinner at %I:%M %p")

print(formatted_dinner)
```
