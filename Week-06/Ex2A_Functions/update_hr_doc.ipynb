## Lab 3. Employee data

### This lab exercise makes updates to information extracted from
### an employee database.

```python
hr_list = [
    ('0123', 'HR', 'Rebecca Yang', '69000'),
    ('0121', 'IT', 'Mark Blick', '67000'),
    ('0068', 'IT', 'Kahna Larsen', '112000'),
    ('0234', 'OPS', 'Jim Smith', '54000')
]
```

```python
new_list = []  # empty list to update

for emp in hr_list:
    if emp[2] == 'Mark Blick':
        # append updated tuple
        new_list.append((emp[0], emp[1], 'Mark Blick-Hawley', emp[3])) # changing the location we're looking for which is adding(append) something to the name 
    else:
        new_list.append(emp)  # append unchanged tuple
```

```python
print(new_list)
```

```python
updated_list = [] # Updating list again

for emp in hr_list:
    if emp[2] == 'Jim Smith': # finding emp in list(tuple) 
        updated_list.append((emp[0], 'CS', emp[2], '60000')) # change department to CS and salary to 60000 and keep everything else the same
    else:
        updated_list.append(emp) # Replace original list so changes continue

hr_list = updated_list
```

```python
print(hr_list)
```

```python
for emp in hr_list:
    print(f"{emp[0]} | {emp[1]} | {emp[2]} | {int(emp[3]):,}")
# using | to format correctly(I looked up how to code this so it looks viually easy to read.
```
