## Lab 2. Parsing strings – part codes

### The values

```python
# Listing out the supplier code, product num, and size for each code, values to variables
code1 = "ACME:123-L"
code2 = "DI:12345-M"
code3 = "ACE:1-12"
```

### Def and Return statements fro the supplier code/num/size

```python
# finding the supplier code before the colon
def get_supplier_code(code):
    for i in range(len(code)):
        if code[i] == ':': 
            return code[:i]
```

```python
# Finding the product num using .find() from whats in between ":" and "-" in my values and variables 
def get_product_num(code):
    colon = code.find(':') + 1 # finding whats after the colon
    dash = code.find('-') # finding whats before the dash 
    return code[colon:dash]
```

```python
# Getting the part of codes that are after the dash
def get_size(code):
    return code.split('-')[1] # 
```

## Adding values to our def/return statements

```python
supplier = get_supplier_code(code1),
product = get_product_num(code1),
size = get_size(code1)

print(f"For {code1}, the supplier is {supplier}. Part number {product} is available in size {size}.") # print format from the workbook 
```

```python
supplier = get_supplier_code(code2),
product = get_product_num(code2),
size = get_size(code2)

print(f"For {code2}, the supplier is {supplier}. Part number {product} is available in size {size}.") # print format from the workbook 
```
