## Exercise 2.C, working with defining custom functions

```python
# Define and return
def favorite_things(name, movie, music):
    return f"Name: {name}\nFavorite Movie: {movie}\nFavorite Music: {music}"

# using print favorite_things
print(favorite_things("Naomi Tewolde", "Everything, Everywhere, All at Once", "Mitski"))
```

```python
# Define and return
def why_im_here(purpose):
    return purpose

# Call the function in a print statement
print(why_im_here("I joined the Year Up United Data Analyst program to build my skills in data and start a strong career in tech."))
```

```python
# Define fav_place
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
print(favorite_place("Japan", "I love the culture, the food, and the beautiful scenery"))
```
