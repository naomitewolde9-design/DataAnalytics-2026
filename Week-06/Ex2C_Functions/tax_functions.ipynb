# Payroll Tax Functions Lab

In this lab exercise, we used Python in jupyter notebook to create functions that calculate different types of payroll taxes based on an employee’s gross pay.

The purpose of this lab is to demonstrate how functions can be used to organize code and perform repeated calculations efficiently using def/return and if/elif/else statements. 

The 3 main functions I created and why:
- A function to calculate Social Security tax (6.2%)
- A function to calculate Medicare tax (1.45%)
- A function to calculate federal withholding tax based on a withholding code and a set of tax rates

Finally, we tested our functions using different codes identification and print() to verify that the calculations are correct.

This lab helps build skills in conditional logic and basic financial calculations in Python/Jupyter notebook.

```python
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062  # 6.2%
```

```python
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145  # 1.45%
```

```python
def get_federal_tax(gross_pay, code):
    if code == 0:
        rate = 0.23
    elif code == 1:
        rate = 0.21
    elif code == 2:
        rate = 0.195
    elif code == 3:
        rate = 0.185
    else:
        # code 4+
        rate = 0.18 - ((code - 4) * 0.005)
    
    return gross_pay * rate
```

```python
gross = 750
code = 0

print("Person 1")
print("Social Security Tax:", get_soc_sec_tax(gross))
print("Medicare Tax:", get_medicare_tax(gross))
print("Federal Tax:", get_federal_tax(gross, code))
```
